## Charger data/train.csv et voir sa structure


```python
import pandas as pd

# Charger uniquement l'ensemble d'entraînement (jamais le test avant l'évaluation finale)
train = pd.read_csv("../data/train.csv")

# Dimensions et aperçu
print(train.shape)
train.head(3)
```

```python
train.describe()
```

## Lister les groupes de variables

```python
# 1. Variables numériques (nombres continus ou comptages / durées / scores)
num_cols = [
    "InfoPages",
    "InfoTimeSec",
    "CatalogPages",
    "CatalogTimeSec",
    "AdminPages",
    "AdminTimeSec",
    "BounceRate",
    "ExitRate",
    "PageValue",
    "EngagementValue",
    "PromoProximity",
]

# 2. Variables catégorielles nominales (texte ou codes sans ordre naturel)
cat_cols = [
    "CampaignType",
    "CustomerType",
    "DeviceOS",
    "BrowserCode",
    "GeoZone",
    "EntryChannel",
]

# 3. Variables binaires (2 valeurs, ex. 0/1 ou booléen)
bin_cols = [
    "Weekend",
]

# Vérifier les modalités des variables catégorielles et binaires
for col in cat_cols + bin_cols:
    if col in train.columns:
        print(f"\n--- {col} ---")
        print(train[col].value_counts(dropna=False))
```

## Garder uniquement les colonnes présentes dans train

```python
num_cols = [c for c in num_cols if c in train.columns]
cat_cols = [c for c in cat_cols if c in train.columns]
bin_cols = [c for c in bin_cols if c in train.columns]

print("Numériques :", len(num_cols), num_cols)
print("Catégorielles :", len(cat_cols), cat_cols)
print("Binaires :", len(bin_cols), bin_cols)
```
